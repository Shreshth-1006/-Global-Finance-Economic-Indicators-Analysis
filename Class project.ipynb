```python
# Load Dataset
import numpy as np
import pandas as pd
df = pd.read_csv("/content/Global finance data.csv")

print('Dataset loaded successfully.')
print(f'  Rows    : {df.shape[0]:,}')
print(f'  Columns : {df.shape[1]}')
print()
df.head()
```

```python
# Missing values
df.isnull().sum()

# Cleaning
df = df.dropna()
df = df.drop_duplicates()
```

## 🌱 4. Economic Growth Analysis — GDP Trends Across Countries

- Compare GDP growth across countries
- Identify highest and lowest growing economies
- Observe patterns in economic growth




**"This helps us understand which economies are expanding rapidly."**

```python
import matplotlib.pyplot as plt

gdp = df.sort_values(by='GDP_Growth_Rate_Percent', ascending=False)

plt.figure(figsize=(14,8))
plt.bar(gdp['Country'], gdp['GDP_Growth_Rate_Percent'])
plt.xticks(rotation=90)
plt.title("GDP Growth Across Countries")
plt.show()
```

##💸 5. Inflation Analysis — Understanding Price Stability
- Compare inflation rates between countries
- Identify countries with high inflation
- Analyze impact of inflation on economy

**“Inflation reflects price stability and economic health.”**

```python
inflation = df.sort_values(by='Inflation_Rate_Percent', ascending=False)

plt.figure(figsize=(14,8))
plt.bar(inflation['Country'], inflation['Inflation_Rate_Percent'])
plt.xticks(rotation=90)
plt.title("Inflation Rate Across Countries")
plt.show()
```

##📈 6. Financial Market Analysis — Stock Index Performance

- Identify top-performing markets

##🏦 7. Monetary Policy Analysis — Interest Rates Impact

##⚠️ 10. Risk Analysis — Political Stability and Financial Risk

##🔗 11. Correlation Analysis — Interdependence of Economic Indicators
- Study relationships between variables
- Identify positive and negative correlations
- Understand how indicators influence each other

**“This reveals hidden relationships in the data.”**

```python
import seaborn as sns
cols = [
    'GDP_Growth_Rate_Percent',
    'Inflation_Rate_Percent',
    'Index_Value',
    'Interest_Rate_Percent',
    'Unemployment_Rate_Percent']

corr = df[cols].corr()

plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

##🌍 12. Comparative Analysis — Developed vs Emerging Economies
- Group countries based on development level
- Compare GDP, inflation, and risk
- Identify differences in economic patterns


**“This helps us understand global economic diversity.”**

```python
import matplotlib.pyplot as plt
top = df.nlargest(5, 'GDP_Growth_Rate_Percent')
bottom = df.nsmallest(5, 'GDP_Growth_Rate_Percent')

print("Top Economies:\n", top[['Country','GDP_Growth_Rate_Percent']])
print("\nWeak Economies:\n", bottom[['Country', 'GDP_Growth_Rate_Percent']])

plt.pie(df.nlargest(5,'GDP_Growth_Rate_Percent')['GDP_Growth_Rate_Percent'],
        labels=df.nlargest(5,'GDP_Growth_Rate_Percent')['Country'],
        autopct='%1.1f%%')
plt.title("Top Economies")
plt.show()

# Changed from pie chart to bar chart to handle negative GDP growth rates
plt.figure(figsize=(10, 6))
plt.bar(df.nsmallest(5,'GDP_Growth_Rate_Percent')['Country'],
        df.nsmallest(5,'GDP_Growth_Rate_Percent')['GDP_Growth_Rate_Percent'],
        color=['red' if x < 0 else 'green' for x in df.nsmallest(5,'GDP_Growth_Rate_Percent')['GDP_Growth_Rate_Percent']])
plt.xlabel("Country")
plt.ylabel("GDP Growth Rate (%)")
plt.title("Weak Economies")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X = df[['GDP_Growth_Rate_Percent', 'GDP_Growth_Rate_Percent', 'Interest_Rate_Percent']]
y = df['Index_Value']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LinearRegression()
model.fit(X_train, y_train)

print("Model Accuracy:", model.score(X_test, y_test))
```
